### 3) Combine all dataframes, check duplicates, generate corresponding csv products

```python
input_dataframe_paths = [temporary_saving_data_dir+"df_species_update_mexico.csv",
                         temporary_saving_data_dir+"df_species_inaturalist2024.csv",
                         temporary_saving_data_dir+"df_species_stavana.csv",
                         temporary_saving_data_dir+"df_species_newAmerica.csv",
                         temporary_saving_data_dir+"df_species_paty.csv",
                         temporary_saving_data_dir+"df_Alejandro.csv",
                         temporary_saving_data_dir+"df_new_collect.csv"]
```

```python
# prepare raw data
frames = []
for this_path in input_dataframe_paths:
    this_frame = pd.read_csv(this_path)
    frames.append(this_frame)

df_species = pd.concat(frames, axis=0)
df_species = df_species.reset_index()

df_species.to_csv("/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/raw_dataset.csv", index=False)
```

```python
combined_america_df, combined_north_america_df = cleaning.combine_all_check_duplicate_and_save_edit2(input_dataframe_paths,final_csvs_saving_dir)
```

```python
combined_america_df.to_csv("/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/cleaned_americas.csv", index=False)
combined_north_america_df.to_csv("/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/cleaned_northamerica.csv", index=False)
combined_america_df['Genus_Species'] = combined_america_df['Genus'] + ' ' + combined_america_df['Species']
combined_america_df['Genus_Species'].value_counts().to_csv('/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/cleaned_americas_count.csv', header=True)
combined_north_america_df['Genus_Species'] = combined_north_america_df['Genus'] + ' ' + combined_north_america_df['Species']
combined_north_america_df['Genus_Species'].value_counts().to_csv('/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/cleaned_northamerica_count.csv', header=True)
df_species['Genus_Species'] = df_species['Genus'] + ' ' + df_species['Species']
df_species['Genus_Species'].value_counts().to_csv("/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/raw_count.csv")
```

```python
#generate csv
import pandas as pd

# File paths
file1 = "/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/cleaned_americas_count.csv"
file2 = "/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/cleaned_northamerica_count.csv"
file3 = "/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/raw_count.csv"

# Load CSV files
df_america = pd.read_csv(file1)
df_north_america = pd.read_csv(file2)
df_species = pd.read_csv(file3)
print(df_species.head())
print(df_america.head())
print(df_north_america.head())


# Merge DataFrames on 'Genus_Species' using an outer join to keep all data
combined_df = df_america.merge(df_north_america, on='Genus_Species', how='outer', suffixes=('_america', '_north_america'))
combined_df = combined_df.merge(df_species, on='Genus_Species', how='outer')

# Save the combined DataFrame to CSV
output_file = "/your_path/Documents/GitHub/Clean_species_data/data/edit_final_data_americas/combined_species_data.csv"
combined_df.to_csv(output_file, index=False)

print(f"Combined species data saved to {output_file}")
```
